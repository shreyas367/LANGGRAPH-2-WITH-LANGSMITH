SEMANTIC SEARCH

```python
from langchain_huggingface import HuggingFaceEmbeddings

embed_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
    encode_kwargs={"normalize_embeddings": True},
)
```

```python
store=InMemoryStore(index={'embed':embed_model,'dims':324})
```

```python
store.put(namespace2, "3",  {"data":"user owns a blue bicycle"});
store.put(namespace2, "4",  {"data":"user wakes up early on weekends"});
store.put(namespace2, "5",  {"data":"user enjoys sci-fi movies"});
store.put(namespace2, "6",  {"data":"user has visited Mumbai"});
store.put(namespace2, "7",  {"data":"user prefers tea over coffee"});
store.put(namespace2, "8",  {"data":"user listens to podcasts"});
store.put(namespace2, "9",  {"data":"user likes rainy weather"});
store.put(namespace2, "10", {"data":"user collects stickers"});
store.put(namespace2, "11", {"data":"user plays chess occasionally"});
store.put(namespace2, "12", {"data":"user enjoys late night coding"});
```

```python
example=store.search(namespace2,query="user like which weather",limit=1)
example
```
